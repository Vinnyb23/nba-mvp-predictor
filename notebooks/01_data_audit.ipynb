# NBA MVP Predictor: Data Audit

## Objective
Explore the historical NBA MVP voting data before building a data-collection and machine-learning pipeline.

## Questions to answer
1. What columns are available in the MVP voting table?
2. Which seasons are present?
3. Is MVP voting available as a numeric modeling target?
4. Which features contain missing values?
5. What data-cleaning issues need to be addressed before modeling?

```python
# Use one completed season for the initial data-collection test.
season_end_year = 2025

# Basketball Reference uses the ending calendar year in this URL.
url = f"https://www.basketball-reference.com/awards/awards_{season_end_year}.html"

# Read every HTML table available on the page.
tables = pd.read_html(url)

print(f"Number of tables found: {len(tables)}")

# Display the column names and first five rows from each table.
for i, table in enumerate(tables):
    print(f"\nTable {i}")
    print(table.columns.tolist())
    display(table.head())
```

```python
# Select the MVP voting table from the list of page tables.
mvp_2025 = tables[0].copy()

# Inspect the structure and the first records.
display(mvp_2025.head())

print("Dataset shape:", mvp_2025.shape)

print("\nColumn names:")
print(mvp_2025.columns.tolist())

print("\nData types and non-null counts:")
mvp_2025.info()
```

```python
# Count missing values in each column.
missing_values = (
    mvp_2025.isna()
    .sum()
    .sort_values(ascending=False)
)

# Show only columns that contain at least one missing value.
display(missing_values[missing_values > 0])
```

```python
print(mvp_2025.columns)
print("\nColumn index type:")
print(type(mvp_2025.columns))
```

```python
# Convert two-level column labels into single, consistent names.
mvp_2025.columns = [
    second_level
    if first_level.startswith("Unnamed")
    else f"{first_level}_{second_level}"
    for first_level, second_level in mvp_2025.columns
]

# Standardize names for Python-friendly use.
mvp_2025.columns = (
    mvp_2025.columns
    .str.lower()
    .str.replace("%", "_pct", regex=False)
    .str.replace("/", "_per_", regex=False)
    .str.replace(" ", "_", regex=False)
)

print(mvp_2025.columns.tolist())
display(mvp_2025.head())
```

```python
# Check duplicate records in this MVP-voting table.
duplicate_rows = mvp_2025.duplicated().sum()

# Check whether any player appears more than once in the same season.
duplicate_players = mvp_2025.duplicated(subset=["player"]).sum()

print(f"Exact duplicate rows: {duplicate_rows}")
print(f"Duplicate player names: {duplicate_players}")

# Review the distribution of numeric features.
display(mvp_2025.describe())
```

```python
# Display MVP candidates in descending order of vote share.
mvp_ranking = (
    mvp_2025[
        [
            "rank",
            "player",
            "tm",
            "voting_first",
            "voting_pts_won",
            "voting_pts_max",
            "voting_share",
            "per_game_pts",
            "per_game_trb",
            "per_game_ast",
            "advanced_ws",
            "advanced_ws_per_48",
        ]
    ]
    .sort_values("voting_share", ascending=False)
    .reset_index(drop=True)
)

display(mvp_ranking)

# Validate the voting-share range.
print("Minimum voting share:", mvp_2025["voting_share"].min())
print("Maximum voting share:", mvp_2025["voting_share"].max())
print("Total voting share:", mvp_2025["voting_share"].sum())
```
